```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.DataFrame({
'x': [1, 2, 3, 4, 5],
'y1': [2, 3.5, 4, 5, 4.5],
'y2': [3, 5, 7, 6, 4],
})
plt.figure(figsize=(10,6))
plt.subplot(2, 2, 1) # so dong, so cot, vi tri
plt.plot(df['x'], df['y1'], color='r')
plt.title('Biểu đồ đường')
plt.subplot(2, 2, 4)
plt.bar(df['x'], df['y2'], color='b')
plt.title('Biểu đồ cột')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt


df = pd.DataFrame({
    'x': [0, 1, 2],
    'a': [2, 3.5, 3],
    'b': [3, 5, 7],
    'c': [1, 3, 8],
    'd': [3, 5, 7],
})


plt.figure(figsize=(10, 6))


plt.subplot(2, 2, 1)
df.plot(kind='line', x='x', y=['a', 'b'],title='Line chart', ax=plt.gca())


plt.subplot(2, 2, 2)
df.plot(kind='bar', x='x', y=['b', 'c'],title='Bar chart', stacked = 'True', ax=plt.gca())


plt.subplot(2, 2, 3)
df.plot(kind='barh', x='x', y=['a', 'd'],title='Horizontal Bar chart', stacked = 'True', ax=plt.gca())


plt.subplot(2, 2, 4)
df.plot(kind='bar', x='x', y='a',title='Mix chart', ax=plt.gca())


plt.subplot(2, 2, 4)
df.plot(kind='line', x='x', y='c',title='Mix chart', ax=plt.gca())


plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt


df = pd.DataFrame({
    'months': ["Jan 19","Feb 19","Mar 19","Apr 19","May 19","Jun 19","Jul 19","Aug 19","Sep 19","Oct 19","Nov 19","Dec 19"],
    'actuals': [1650,1600,1450,1200,1080,820,700,900,1250,0,0,0],
    'forecast': [1700,1650,1300,1200,1100,900,750,850,1350,1550,1750,2000]
})


plt.figure(figsize=(20, 10))


plt.subplot(2,2,1)
df.plot(x='months', y='actuals', kind='bar', title='Units Sold in Canada - Actuals', color='#55C2B1', ax=plt.gca())


plt.subplot(2,2,1)
df.plot(x='months', y='forecast', kind='line', title='Units Sold in Canada - Forecast', color='#B9382D', marker = 'o', ax=plt.gca())
plt.ylim(0, 2500)
plt.grid(axis='y', alpha=0.25)
plt.xlabel('')
plt.ylabel('Units Sold')
plt.legend(['Forecast', 'Actual'], loc='upper center', bbox_to_anchor=(0.5, -0.08), ncol=2, frameon=False)
plt.tight_layout()
plt.show()
```
